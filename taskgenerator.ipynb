```python
import pandas as pd
import random
from faker import Faker
```

```python
#Total rows needed
fake = Faker()
num_tasks = 1500
```

```python

priorities = ["Low", "Medium", "High", "Critical"]
statuses = ["Pending", "In Progress", "Completed", "On Hold"]
categories = ["Bug Fix", "Feature", "UI/UX", "Testing", "Documentation"]
assignees = ["Shan", "Jo", "Jane", "Dia", "Akin"]
```

```python
data = []

for i in range(num_tasks):
    data.append({
        "task_id": i + 1,
        "task_title": fake.sentence(nb_words=4),
        "task_description": fake.text(max_nb_chars=120),
        "priority": random.choice(priorities),
        "deadline": fake.date_between(start_date="+1d", end_date="+30d"),
        "assigned_to": random.choice(assignees),
        "status": random.choice(statuses),
        "category": random.choice(categories),
        "estimated_hours": round(random.uniform(1, 10), 1),
        "created_at": fake.date_time_this_month()
    })
```

```python

df = pd.DataFrame(data)
df.head(3)
```

```python
df.to_csv("synthetic_task_dataset.csv", index=False)
print("✅ Synthetic dataset created successfully!")
```
